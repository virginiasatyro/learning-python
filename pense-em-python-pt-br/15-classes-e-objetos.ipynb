## Tipos definidos pelos progamadores

Um tipo definido pelo programador também é chamado de **classe**.

```python
class Point:
  """ represents a point in 2D space """
```

Definir uma classe denominada `Point`cria um **objeto de classe**:

```python
Point
```

```python
Point()
```

```python
blank = Point()
blank
```

O valor de retorno é uma referência a um objeto `Point`, ao qual atribuímos `blank`.
Criar um objeto chama-se **instanciação**, e o objeto é uma **instância** da classe.

## Atributos

Você pode atribuir atributos a um objeto usando a notação de ponto:

```python
blank.x = 3.0
blank.y = 4.0
```

Nesse caso, estamos atribuindo valores a elementos nomeados de um objeto. Esses elementos são chamados de **atributos**.

```python
# não há conflito entre variável x e atributo x
x = blank.x
x
```

```python
import math
distance = math.sqrt(blank.x ** 2 + blank.y ** 2)
distance
```

```python
# passando uma instâmcia como argumento
def print_point(p):
  print(f'({p.x}, {p.y})')

print_point(blank)
```

## Retângulos

```python
class Rectangle:
  """ represents a rectangle. 
    attributes: width, height, corner.
  """
```

```python
box = Rectangle()
box.width = 100.0
box.height = 200.0
box.corner = Point()
box.corner.x = 0.0
box.corner.y = 0.0
```

## Instâncias como argumentos

```python
def find_center(rect):
  p = Point()
  p.x = rect.corner.x + rect.width / 2
  p.y = rect.corner.y + rect.height / 2
  return p

center = find_center(box)
print_point(center)
```

## Cópia

```python
p1 = Point()
p1.x = 3.0
p1.y = 4.0
```

```python
import copy
p2 = copy.copy(p1)

print_point(p1)
print_point(p2)
p1 is p2
```

```python
p1 == p2
```

```python
box2 = copy.copy(box)
box2.corner is box.corner
```

Cuidado! `copy.copy` faz uma cópia rasa, o que significa que ele copia o objeto, mas não os objetos aos quais ele se refere. `copy.deepcopy` faz uma cópia profunda, o que significa que ele copia o objeto e os objetos aos quais ele se refere.

```python
box2.corner is box.corner
```

```python
box3 = copy.deepcopy(box)
box3.corner is box.corner
```

```python
box3.corner is box.corner
```

## Glossário

- **classe**
  - Tipo definido pelo programador. Uma definição de classe cria um objeto de classe.

- **objeto de classe**
  - Objeto que contém a informação sobre um tipo definido pelo programador. O objeto de classe pode ser usado para criar instâncias do tipo.

- **instancia**
  - Obejto que pertence a uma classe.

- **instanciar**
  - Criar um objeto.

- **atributo**
  - Um dos valores denominados associados a um objeto.

- **objeto integreado**
  - Objeto que é armazenado como um atributo de outro objeto.

- **cópia superficial**
  - Cópiar o conteúdo de um objeto, inclusive qualquer referência a outros obejetos integrados; implementada pela função `copy` no módulo `copy`.

- **cópia profunda**
  - Copiar o conteúdo de um objeto, bem como qualquer objeto integrado, e qualquer objeto integrado a estes, e assim por diante; implementado pela função `deepcopy` no módulo `copy`.

- **diagrama de classes**
  - Diagrama que mostra objetos, seus atributos e os valores dos atributos.

Escreva uma definição para uma classe denominada `Circle`, com os atributos center e radius, onde center é um objeto Point e radius é um número.

Instancie um objeto `Circle`, que represente um círculo com o centro em 150, 100 e raio 75.

Escreva uma função denominada `point_in_circle`, que tome um `Circle` e um `Point` e retorne `True`, se o ponto estiver dentro ou no limite do círculo.

Escreva uma função chamada `rect_in_circle`, que tome um `Circle` e um `Rectangle` e retorne `True`, se o retângulo estiver totalmente dentro ou no limite do círculo.

Escreva uma função denominada `rect_circle_overlap`, que tome um `Circle` e um `Rectangle` e retorne `True`, se algum dos cantos do retângulo cair dentro do círculo. Ou, em uma versão mais desafiadora, retorne `True` se alguma parte do retângulo cair dentro do círculo.

Solução: http://thinkpython2.com/code/Circle.py.


```python
class Circle:
  """ represents a circle. 
    attributes: radius, center.
  """

c = Circle()
c.radius = 75.0
c.center = Point()
c.center.x = 150.0
c.center.y = 100.0
```

```python
def point_in_circle(circle, point):
  """ returns True if the point is inside the circle. """
  distance = math.sqrt((point.x - circle.center.x) ** 2 + (point.y - circle.center.y) ** 2)
  return distance < circle.radius
```

```python
def rect_in_circle(circle, rect):
  """ returns True if the rectangle is entirely in the circle. """
  p = find_center(rect)
  distance = math.sqrt((p.x - circle.center.x) ** 2 + (p.y - circle.center.y) ** 2)
  return distance + math.sqrt((rect.width / 2) ** 2 + (rect.height / 2) ** 2) < circle.radius
```

```python
def rect_circle_overlap(circle, rect):
  """ returns True if any of the corners of the rectangle are in the circle. """
  p = find_center(rect)
  distance = math.sqrt((p.x - circle.center.x) ** 2 + (p.y - circle.center.y) ** 2)
  return distance < circle.radius + math.sqrt((rect.width / 2) ** 2 + (rect.height / 2) ** 2)
```

## Exercício 15.2

Escreva uma função chamada `draw_rect` que receba um objeto Turtle e um Rectangle e use o Turtle para desenhar o retângulo. Veja no Capítulo 4 os exemplos de uso de objetos Turtle.

Escreva uma função chamada `draw_circle`, que tome um Turtle e um Circle e desenhe o círculo.

```python
# see exercicio-15.2.1.py
def draw_rect(t, rect):
  """ Draws a Rectangle using turtle t. """
  for i in range(2):
    t.forward(rect.width)
    t.left(90)
    t.forward(rect.height)
    t.left(90)
```

```python
def draw_circle(t, circle):
  """ Draws a Circle using turtle t. """
  t.up()
  t.forward(circle.radius)
  t.left(90)
  t.down()
  t.circle(circle.radius)
```
